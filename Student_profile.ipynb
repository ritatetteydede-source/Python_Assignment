```python
# Step 1: Collect user information

name = input("Enter your full name: ")
age = int(input("Enter your age: "))
gpa = float(input("Enter your GPA (0.0 - 4.0): "))
courses_input = input("Enter your courses (comma-separated): ")
courses = [course.strip() for course in courses_input.split(",")]

# Step 2: Store in a dictionary

student_profile = {
    "name": name,
    "age": age,
    "gpa": gpa,
    "courses": courses,
    "num_courses": len(courses)
}

# Step 3: Display formatted output

deans_list = gpa >= 3.5
is_minor = age < 18

print("\n==== STUDENT PROFILE ====")
print(f"Name: {student_profile['name']}")
print(f"Age: {student_profile['age']}")
print(f"GPA: {student_profile['gpa']}")
print(f"Courses: {', '.join(student_profile['courses'])}")
print(f"Number of Courses: {student_profile['num_courses']}")
print(f"Dean's List: {deans_list}")


if is_minor:
    print("Status: Minor")
else:
    print("Status: Adult")

# Step 4: Show data types

print("\n==== DATA TYPES ====")
print(f"Name = {type(name).__name__}")
print(f"Age = {type(age).__name__}")
print(f"GPA = {type(gpa).__name__}")
print(f"Courses = {type(courses).__name__}")
print(f"Dean's List = {type(deans_list).__name__}")
print(f"Student Profile = {type(student_profile).__name__}")

print("\nComplete Assignment 1 by implementing the student profile system.")
```
